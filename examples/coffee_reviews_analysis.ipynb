## Step 2: Load the Coffee Reviews Dataset

In Kaggle, datasets are automatically mounted in the `/kaggle/input/` directory. Let's set up our workspace and import the coffee reviews data.

### Understanding the Data Structure

The dataset contains:
- **Text data**: Detailed coffee reviews with flavor descriptions
- **Numeric data**: Ratings (84-97), prices ($/100g), and metadata
- **Categorical data**: Origin countries, roast levels, roaster locations

````python
import os
import subprocess
import sys
import pandas as pd
````

````python
# Add The Coffee Reviews Dataset on the "right" side of the screen where it says "Add Input"
! ls -a /kaggle/input/datasets/schmoyote/coffee-reviews-dataset
````

````python
# Import the coffee reviews into a CRISP-T corpus
# The 'review' column contains the text data we'll analyze
# We'll save the corpus to 'workspace/coffee_corpus'

print("🔄 Importing coffee reviews into CRISP-T corpus...")
print("This may take a moment as CRISP-T processes the text data...\n")

# Use CRISP-T's crisp command to import the data
# Using uv run to ensure we use the crisp-t package from our uv environment
# --source: specifies the source directory containing CSV files
# --unstructured: specifies which column contains free-text data
# --out: output directory for the corpus
result = subprocess.run(
    [
        "uv",
        "run",
        "--with",
        "crisp-t[ml,xg]",
        "crisp",
        "--source",
        "/kaggle/input/datasets/schmoyote/coffee-reviews-dataset/",
        "--unstructured",
        "review",
        "--out",
        "workspace/coffee_corpus",
    ],
    capture_output=True,
    text=True,
)

print(result.stdout)
if result.stderr:

    print("⚠️ Warnings/Notes:")
    print(result.stderr)
````

## Step 3: Explore the Data Structure

Now let's examine what we've loaded. CRISP-T creates a corpus that combines both the text documents and the numerical data.

````python
# Check corpus information
print("📊 Loading corpus information...\n")

result = subprocess.run([
    'uv', 'run', '--with', 'crisp-t[ml,xg]', 'crisp',
    '--inp', 'workspace/coffee_corpus',
    '--print', 'documents'
], capture_output=True, text=True)

print(result.stdout)
````

````python
# Let's also check the numerical data structure
print("🔍 Examining numerical data columns...\n")

result = subprocess.run([
    'uv', 'run', '--with', 'crisp-t[ml,xg]', 'crisp',
    '--inp', 'workspace/coffee_corpus',
    '--query', 'head()'
], capture_output=True, text=True)

print(result.stdout)
````

## Step 4: Analyze Rating Statistics

Let's understand the distribution of coffee ratings. This will give us insights into the overall quality profile of the dataset.

### Descriptive Statistics

The mean rating is calculated as: $\mu = \frac{1}{n}\sum_{i=1}^{n} x_i$

````python
# Get descriptive statistics for all numerical columns
print("📈 Computing descriptive statistics...\n")

result = subprocess.run([
    'uv', 'run', '--with', 'crisp-t[ml,xg]', 'crisp',
    '--inp', 'workspace/coffee_corpus',
    '--query', 'describe()'
], capture_output=True, text=True)

print(result.stdout)

print("\n💡 Key Insights:")
print("   • The dataset contains ratings from 84-97 (specialty coffee range)")
print("   • Mean rating around 93.3 indicates high-quality coffees")
print("   • Price varies significantly ($0.17 - $132.28 per 100g)")
````

## Step 5: Geographic Quality Analysis

Let's explore which coffee origins deliver the highest ratings. This is crucial for understanding quality patterns across different regions.

````python
# Analyze ratings by coffee origin
print("🌍 Analyzing average ratings by coffee origin...\n")
print("This reveals which regions produce the highest-rated coffees!\n")

result = subprocess.run([
    'uv', 'run', '--with', 'crisp-t[ml,xg]', 'crisp',
    '--inp', 'workspace/coffee_corpus',
    '--query', "groupby('origin')['rating'].agg(['mean', 'count']).sort_values('mean', ascending=False).head(10)"
], capture_output=True, text=True)

print(result.stdout)

print("\n🏆 Top Performing Origins:")
print("   • Look for origins with both high ratings AND substantial sample sizes")
print("   • This indicates consistent quality, not just lucky outliers")
````

## Step 6: Roast Level Impact on Quality

How does roast level affect ratings? Let's investigate whether lighter or darker roasts receive higher scores.

````python
# Analyze ratings by roast level
print("🔥 Analyzing ratings by roast level...\n")
print("Does roast darkness correlate with quality scores?\n")

result = subprocess.run([
    'uv', 'run', '--with', 'crisp-t[ml,xg]', 'crisp',
    '--inp', 'workspace/coffee_corpus',
    '--query', "groupby('roast')['rating'].agg(['mean', 'count']).sort_values('mean', ascending=False)"
], capture_output=True, text=True)

print(result.stdout)

print("\n💡 Roast Insights:")
print("   • Lighter roasts typically preserve more origin characteristics")
print("   • Darker roasts emphasize roast-derived flavors")
print("   • Specialty coffee often favors lighter roasts for complexity")
````

## Step 7: Price-Quality Relationship

One of the most interesting questions: **Does higher price mean better quality?**

Let's examine how average price varies by rating level.

````python
# Analyze average price by rating level
print("💰 Examining price vs. quality relationship...\n")
print("How does price change as ratings increase?\n")

result = subprocess.run([
    'uv', 'run', '--with', 'crisp-t[ml,xg]', 'crisp',
    '--inp', 'workspace/coffee_corpus',
    '--query', "groupby('rating')['100g_USD'].mean().sort_index()"
], capture_output=True, text=True)

print(result.stdout)

print("\n💡 Price-Quality Insights:")
print("   • Notice how prices jump significantly at rating 95+")
print("   • Premium coffees (96-97) command 2-3x the average price")
print("   • This suggests diminishing returns at the highest quality levels")
````

## Step 8: Textual Analysis - Topic Modeling

Now let's dive into the **textual content** of the reviews! We'll use **Latent Dirichlet Allocation (LDA)** to discover hidden thematic patterns in how people describe coffee flavors.

### What is Topic Modeling?

Topic modeling automatically discovers the main themes in a collection of documents. Each topic is represented as a combination of words that frequently appear together.

````python
# Run topic modeling with 10 topics
print("🎯 Discovering thematic patterns in coffee reviews...\n")
print("This will identify the main flavor themes across all reviews.\n")
print("⏳ Topic modeling may take 1-2 minutes...\n")

result = subprocess.run([
    'uv', 'run', '--with', 'crisp-t[ml,xg]', 'crisp',
    '--inp', 'workspace/coffee_corpus',
    '--out', 'workspace/coffee_corpus',
    '--topics',
    '--num', '10'
], capture_output=True, text=True)

print(result.stdout)

if result.stderr:
    # Filter out warnings, show important info
    error_lines = [line for line in result.stderr.split('\n') if 'INFO' in line]
    if error_lines:
        print("\nℹ️ Additional info:")
        for line in error_lines[:5]:
            print(line)
````

## Step 9: Sentiment Analysis

Let's analyze the **emotional tone** of the coffee reviews. We'll use VADER (Valence Aware Dictionary and sEntiment Reasoner) to understand the sentiment profile.

````python
# Run sentiment analysis on coffee reviews
print("😊 Analyzing sentiment in coffee reviews...\n")
print("VADER will measure positive, neutral, and negative sentiment.\n")

result = subprocess.run([
    'uv', 'run', '--with', 'crisp-t[ml,xg]', 'crisp',
    '--inp', 'workspace/coffee_corpus',
    '--out', 'workspace/coffee_corpus',
    '--sentiment'
], capture_output=True, text=True)

print(result.stdout)

print("\n💡 Sentiment Insights:")
print("   • Specialty coffee reviews tend to be overwhelmingly positive")
print("   • High neutral scores indicate descriptive, technical language")
print("   • Compound score near 1.0 suggests very positive overall tone")
````

## Step 10: Category Analysis - Key Descriptors

Let's extract the most important **descriptive terms** used in coffee reviews. This reveals the vocabulary of specialty coffee tasting.

````python
# Extract key categories and themes from reviews
print("🏷️ Extracting key descriptive categories...\n")
print("This identifies the most important terms used to describe coffee.\n")

result = subprocess.run([
    'uv', 'run', '--with', 'crisp-t[ml,xg]', 'crisp',
    '--inp', 'workspace/coffee_corpus',
    '--cat',
    '--rec', '20'
], capture_output=True, text=True)

print(result.stdout)

print("\n💡 Category Insights:")
print("   • 'Sweet' is the most common descriptor - universal in quality coffee")
print("   • 'Cup' and 'aroma' are central sensory dimensions")
print("   • These terms form the foundation of coffee tasting vocabulary")
````

## Step 11: Clustering Analysis - Market Segments

Let's use **K-Means clustering** to identify distinct segments in the coffee market based on price and rating patterns.

````python
# Run K-Means clustering to identify market segments
print("🔍 Identifying market segments through clustering...\n")
print("This will group coffees by similar price-rating profiles.\n")

result = subprocess.run([
    'uv', 'run', '--with', 'crisp-t[ml,xg]', 'crisp',
    '--inp', 'workspace/coffee_corpus',
    '--out', 'workspace/coffee_corpus',
    '--kmeans'
], capture_output=True, text=True)

# Extract and display key clustering results
output_lines = result.stdout.split('\n')
for i, line in enumerate(output_lines):
    if 'Cluster:' in line or 'Cluster Length' in line or 'Centroids' in line:
        # Print the cluster info and next few lines
        print(line)
        if i + 1 < len(output_lines):
            print(output_lines[i + 1])
        if i + 2 < len(output_lines) and '100g_USD' in output_lines[i + 2]:
            print(output_lines[i + 2])
            print(output_lines[i + 3])

print("\n💡 Clustering Insights:")
print("   • Three distinct segments emerge: Mainstream, Premium, Ultra-Premium")
print("   • Mainstream (~85%): Accessible specialty coffees")
print("   • Premium (~14%): High-end with exceptional quality")
print("   • Ultra-Premium (~2%): Rare luxury coffees (Geisha, microlots)")
````

## Step 12: Regression Analysis - Predicting Coffee Ratings

Now let's use **linear regression** to quantify the relationship between coffee ratings and price. This will help us understand how much price influences quality scores.

### What is Linear Regression?

Linear regression models the relationship between a dependent variable (rating) and one or more independent variables (price). The model finds the best-fitting line: $\hat{y} = \beta_0 + \beta_1 x_1 + \epsilon$

Where:
- $\hat{y}$ = predicted rating
- $\beta_0$ = intercept (base rating)
- $\beta_1$ = coefficient (price effect)
- $x_1$ = price (100g_USD)

````python
# Run linear regression to predict ratings from price
print("📊 Running linear regression analysis...\n")
print("Predicting coffee ratings based on price (100g_USD).\n")
print("⏳ Regression analysis may take 30-60 seconds...\n")

result = subprocess.run([
    'uv', 'run', '--with', 'crisp-t[ml,xg]', 'crisp',
    '--inp', 'workspace/coffee_corpus',
    '--out', 'workspace/coffee_corpus',
    '--regression',
    '--outcome', 'rating',
    '--include', '100g_USD'
], capture_output=True, text=True)

print(result.stdout)

if result.stderr:
    # Filter out warnings, show important info
    error_lines = [line for line in result.stderr.split('\n') if 'INFO' in line or 'WARNING' in line]
    if error_lines:
        print("\nℹ️ Additional info:")
        for line in error_lines[:3]:
            print(line)

print("\n💡 Regression Insights:")
print("   • R² score shows how well price explains rating variance")
print("   • Positive coefficient means higher prices correlate with higher ratings")
print("   • Intercept represents base rating when price is zero")
print("   • This quantifies the price-quality relationship we observed earlier!")
````

## Step 13: Neural Network Analysis - Classifying Roast Levels

Let's use a **neural network** to classify roast levels based on price and rating. Neural networks can capture complex, non-linear patterns that simple regression might miss.

### What is a Neural Network?

A neural network is a machine learning model inspired by biological neural networks. It consists of:
- **Input layer**: Features (price, rating)
- **Hidden layers**: Learn complex patterns through non-linear transformations
- **Output layer**: Predicted class (roast level)

The network learns by adjusting weights to minimize prediction error.

````python
# Run neural network classification to predict roast levels
print("🧠 Training neural network for roast classification...\n")
print("Predicting roast level based on price and rating.\n")
print("⏳ Neural network training may take 1-2 minutes...\n")

result = subprocess.run([
    'uv', 'run', '--with', 'crisp-t[ml,xg]', 'crisp',
    '--inp', 'workspace/coffee_corpus',
    '--out', 'workspace/coffee_corpus',
    '--nnet',
    '--outcome', 'roast',
    '--include', '100g_USD,rating'
], capture_output=True, text=True)

print(result.stdout)

if result.stderr:
    # Filter out warnings, show important info
    error_lines = [line for line in result.stderr.split('\n') if 'INFO' in line]
    if error_lines:
        print("\nℹ️ Training info:")
        for line in error_lines[:3]:
            print(line)

print("\n💡 Neural Network Insights:")
print("   • Accuracy shows how well price+rating predict roast level")
print("   • Higher accuracy means roast level is strongly tied to market positioning")
print("   • Lower accuracy suggests roast is independent of price/rating")
print("   • Neural networks can capture complex patterns beyond linear relationships!")
````
